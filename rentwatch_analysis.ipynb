```python
datasets = {}

for file in Path(".").rglob("*.csv"):

    if any(code in file.name for code in ["B25064", "B25119", "B25070"]):

        df = pd.read_csv(file)

        datasets[file.name] = df

print("Datasets loaded:", len(datasets))

print(datasets.keys())
```

```python
selected_datasets = {}

for name, df in datasets.items():

    print("\nDATASET:", name)

    if "B25064" in name:
        selected = df[
            df["Label (Grouping)"] == "Median gross rent"
        ]

    elif "B25119" in name:
        selected = df[
            df["Label (Grouping)"].str.strip() == "Renter occupied (dollars)"
        ]

    elif "B25070" in name:
        selected = df[
            df["Label (Grouping)"].str.strip().isin([
                "Total:",
                "50.0 percent or more",
                "Not computed"
            ])
        ]

    selected_datasets[name] = selected

    display(selected)
```

```python
import pandas as pd

rent = pd.read_csv("cleaned_median_gross_rent_2024.csv")
income = pd.read_csv("cleaned_renter_household_income_2024.csv")
burden = pd.read_csv("cleaned_severe_rent_burden_2024.csv")

print("MEDIAN GROSS RENT")
display(rent.head())

print("RENTER HOUSEHOLD INCOME")
display(income.head())

print("SEVERE RENT BURDEN")
display(burden.head())
```

```python
def to_number(value):
    """Turn text like '1,678' or '±23' into a real number."""
    if pd.isna(value):
        return None
    text = str(value)
    for junk in [",", "±", "$", "+"]:
        text = text.replace(junk, "")
    try:
        return float(text.strip())
    except ValueError:
        return None
```

```python
records = []
for col in rent.columns:
    if col.endswith("!!Estimate"):
        # "Barnstable County, Massachusetts!!Estimate" -> "Barnstable County"
        county = col.split(", Massachusetts")[0]
        moe_col = f"{county}, Massachusetts!!Margin of Error"
        records.append({
            "County": county,
            "Median Gross Rent": to_number(rent[col].iloc[0]),
            "Rent MOE": to_number(rent[moe_col].iloc[0]),
        })
```

```python
rent_clean = pd.DataFrame(records)
print(rent_clean.shape)
print(rent_clean)
```

```python
print(income.columns.tolist())
print(income.shape)
print(income.iloc[10:18])   # rows around the point where estimates might end
```

```python
values = income.iloc[:, 0].apply(to_number)

estimates = values.iloc[:14].reset_index(drop=True)   # rows 0-13
margins   = values.iloc[14:].reset_index(drop=True)   # rows 14-27

income_clean = pd.DataFrame({
    "County": rent_clean["County"],     # same alphabetical order as the Census files
    "Renter Household Income": estimates,
    "Income MOE": margins,
})
```

```python
print(income_clean.columns.tolist())
# Prints the exact column names that income_clean has right now.
# Compare them letter by letter with "Renter Household Income" and "Income MOE".

print(income_clean.shape)
# Prints (rows, columns), so you can tell if this is the new 14 x 3 table or an old one.

print(income_clean.head())
# Shows the first 5 rows so you can see what the table actually holds.
```

```python
income_clean = income_clean.rename(columns={
    # rename() changes column names. The dictionary reads "old name": "new name".
    "Estimate": "Renter Household Income",
    # The estimate column now says what it actually measures.
    "Margin of Error": "Income MOE",
    # Short, clear name for the margin of error.
})
```

```python
income_clean["County"] = income_clean["County"].str.strip() + " County"
# .str.strip() removes any stray spaces around the name, so " Essex" becomes "Essex".
# + " County" adds the word County to the end: "Essex" becomes "Essex County".
# Your rent and burden tables use "Essex County", so this makes all three agree.
```

```python
print(income_clean.columns.tolist())
# Expect ['County', 'Renter Household Income', 'Income MOE'].

print(income_clean.head())
# Expect names like "Barnstable County" and the same numbers as before.

assert income_clean.shape == (14, 3)
# Stops with an error if the table isn't 14 rows by 3 columns.

assert (income_clean["Renter Household Income"] > income_clean["Income MOE"]).all()
# Every margin of error should be smaller than its estimate.
# If this line prints nothing, that means it passed.
```

```python
print(rent_clean.shape, income_clean.shape, burden.shape)
# Prints the (rows, columns) of each table.
# You want (14, 3), (14, 3) and (14, 2).

print(set(rent_clean["County"]) == set(income_clean["County"]))
# set() turns the county names into an unordered collection of unique names.
# == asks "are the two collections exactly the same?"
# You want True: both tables use the same 14 county names.

print(set(rent_clean["County"]) == set(burden["County"]))
# Same check between rent and severe burden. You want True again.
```

```python
print(rent_clean.shape, income_clean.shape, burden.shape)
# .shape gives (rows, columns) for each table.
# You want (14, 3), (14, 3) and (14, 2).

print(rent_clean.columns.tolist())
print(income_clean.columns.tolist())
print(burden.columns.tolist())
# Shows the column names of each table so you can see exactly what you have.
# Expected:
#   ['County', 'Median Gross Rent', 'Rent MOE']
#   ['County', 'Renter Household Income', 'Income MOE']
#   ['County', 'Severe Rent Burden (%)']
```

```python
for table in [rent_clean, income_clean, burden]:
    # Loops over the three tables one at a time.
    table["County"] = table["County"].str.strip()
    # .str.strip() removes spaces at the start and end of each county name.
```

```python
print(set(rent_clean["County"]) == set(income_clean["County"]))
print(set(rent_clean["County"]) == set(burden["County"]))
# set() turns the names into a collection of unique names.
# == asks "are the two collections identical?"
# You want True twice.
```

```python
merged = rent_clean.merge(income_clean, on="County", how="inner")
# merge() joins two tables side by side.
# on="County" matches rows that have the same county name.
# how="inner" keeps only counties that appear in BOTH tables.
# Result: County, Median Gross Rent, Rent MOE, Renter Household Income, Income MOE.

merged = merged.merge(burden, on="County", how="inner")
# Joins the third table onto the result the same way.
# This adds the Severe Rent Burden (%) column.

print(merged.shape)
# You want (14, 6).

print(merged)
# Shows all 14 rows so you can look them over.
```

```python
assert merged.shape[0] == 14
# shape[0] is the row count. This stops with an error if a county was lost.

assert merged["County"].is_unique
# is_unique is True when no county appears twice.
# This catches duplicates from a bad merge.

print(merged.isna().sum())
# isna() marks missing cells and sum() counts them for each column.
# You want 0 everywhere.

print(merged.describe().round(1))
# Shows min, max and average for each numeric column.
# Sanity check: rents roughly 1,000-2,300, incomes roughly 25,000-50,000,
# severe burden roughly 20-30 percent.
```
